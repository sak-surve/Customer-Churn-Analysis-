```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv(r'customer_churn.csv')
df.head()
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create a 3x3 grid of subplots
fig, axes = plt.subplots(3, 3, figsize=(10,9))
axes = axes.flatten()

# Generate countplots
for i, col in enumerate(cols):
    sns.countplot(
        data=df,
        x=col,
        hue='Churn',
        ax=axes[i]
    )

    # Title and formatting
    axes[i].set_title(f'Churn by {col}')
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].set_xlabel('')

    # Keep legend only on the first subplot
    if i != 0:
        axes[i].get_legend().remove()

# Adjust layout
plt.tight_layout()

# Optional: place a single shared legend outside the grid
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels,
    title='Churn',
    loc='upper right',
    bbox_to_anchor=(1.05, 1)
)

plt.show()
```

```python
"""The majority of customers who do not churn tend to have services like PhoneService,
InternetService (particularly DSL),and OnlineSecurity enabled. For services like OnlineBackups, TechSupport, and StreamingTV,
churn rates are noticeably higher when these srvices are not used or are unavailable. """
```
